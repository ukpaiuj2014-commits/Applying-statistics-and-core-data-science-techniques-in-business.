## Bag-of-Words representation

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
docs = [
        'This is the first project.',
        'This project is the second project., the project is nice',
        'And this is the third one.',
        'Is this the first project?', ]
```

```python
vectorizer = CountVectorizer(analyzer = "word",
                             lowercase=True,
                             max_features = 50)
```

```python
# Convert the documents into a document-term matrix.
wm = vectorizer.fit_transform(docs)
print(wm.todense())
```

Words tranformed in alphabetical order.

```python
vocabulary = vectorizer.vocabulary_
print(vocabulary)
```

Index of each word is highlighted

```python
tokens = vectorizer.get_feature_names_out()
print(tokens)
```

Each word highlighted according to their index, following their alphabetical order

```python
# Create an index for each row.
doc_names = ['Doc{:d}'.format(idx) for idx, _ in enumerate(wm)]
df_bow = pd.DataFrame(data=wm.toarray(), index=doc_names,
                  columns=tokens)
```
