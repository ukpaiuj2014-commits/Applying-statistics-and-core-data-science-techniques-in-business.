# Feature engineering.

```python
# Convert currency to numerical values by removing $
data['Total Revenue'] = data['Total Revenue'].replace(r'[\$,()]','',regex=True).astype(float)
data['Unit Cost'] = data['Unit Cost'].replace(r'[\$,()]','',regex=True).astype(float)
data['Profit'] = data['Profit'].replace(r'[\$,()]','',regex=True).astype(float)

# Convert date columns to datetime objects
data['Order_Date'] = pd.to_datetime(data['Order_Date'], format='%d%b%Y')
data['Delivery_Date'] = pd.to_datetime(data['Delivery_Date'], format='%d%b%Y')
data['Customer_BirthDate'] = pd.to_datetime(data['Customer_BirthDate'], format='%d%b%Y')

print(data.shape)
data.head()
```

```python
# Aggregate the data into one customer per row with the engineered features.
TODAY = pd.to_datetime('today').normalize()
customer_data = data.groupby('Customer ID').agg(

    Recency=('Delivery_Date', lambda x: (TODAY - x.max()).days),
    Frequency=('Order ID', 'count'),
    CLV=('Total Revenue', 'sum'),
    Avg_unit_cost=('Unit Cost', 'mean'),
    Age=('Customer_BirthDate', lambda x: (TODAY - x.max()).days // 365)
).reset_index().set_index('Customer ID')

print(customer_data.shape)
customer_data.head()
```

**Observations from Feature Engineering:**
- For effective feature engineering procedure, all inconsistent data (currencies and dates) were harmonised.
- The dataFrame were aggregated into 5 feature and grouped into one customer per row using their unique Customer ID.

- IQR

```python
# Detect Outliers using IQR plots.
def detect_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers

print("Outlier Detection Summary:")
for col in features:
    col_outliers = detect_outliers(customer_data, col)
    print(f"\n--- Column: {col} ---")
    if not col_outliers.empty:
        print(f"Number of outliers detected: {len(col_outliers)}")
    else:
        print("No outliers detected:")
```

# Clustering with ML Algorithms

Feature Scaling:

```python
# Normalise Dataset using z-score standardization.
scaler = StandardScaler()
scaled_data = scaler.fit_transform(customer_data)

# View scaled Dataset in a DataFrame
scaled_data = pd.DataFrame(scaled_data, columns=customer_data.columns)
print(scaled_data.shape)
scaled_data.head()
```

### Elbow Method.

```python
# Plot Elbow Curve: Generate a line plot with k on the x-axis and WCSS on the y-axis (with Scaled Data).
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(scaled_data)
    wcss.append(kmeans.inertia_)

    # Plot the Elbow Method results
plt.figure(figsize=(10, 10))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```

```python
# Perform Elbow method to optimise k with Unscaled Data.
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(customer_data)
    wcss.append(kmeans.inertia_)

    # Plot the Elbow curve
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method (Unscaled Data)')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

### Silhouette score Method

```python
#Silhouette score Method
silhouette_scores = []
for i in range(2, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)

    cluster_labels = kmeans.fit_predict(scaled_data)
    silhouette_avg = silhouette_score(scaled_data, cluster_labels)
    silhouette_scores.append(silhouette_avg)
    print(f"For n_clusters={i}, the silhouette score is {silhouette_avg}")

    # Plot the silhouette scores
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.title('Silhouette Score Method')
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette Score')
plt.show()
```

### Dendrogram Method

```python
# Perform Hierarchical Clustering with sampled data (n=30000)to create Dendrogram.
small_df = scaled_data.sample(n=30000).reset_index(drop = True)

dendrogram = sch.dendrogram(sch.linkage(small_df, method='ward'),truncate_mode='lastp', p=20, leaf_rotation=90, leaf_font_size=12)
plt.axhline(y=130, color='red', linestyle='--', label='Cut at distance ≈ 130')
plt.legend()

plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean distances')
plt.show()
```

**Observations from determining Optimal number of Clusters (k):**
- To perform any of these methods, the datas et was first scaled because each of the method uses scaled data.
- In Elbow method, unscaled data was used on the first plot, with y-axis(WCss) on log scale. The output suggest a clearer optimal k =5. However Elbow method plot with Scaled data doesnt not give a very clear k, but suggest an opimal k= 6.

- In Silhouette score method, scaled data were used, and based on the silhouette score analysis, the optimal number of clusters for customer segmentation is k = 5. This k value is considered optimal because it yielded the highest silhouette score of approximately 0.2671 among the tested range of 2 to 10 clusters. A higher silhouette score indicates that the clusters are well-separated from other clusters and the data points within a cluster are dense.
- In the Dendrogram, we sampled the scaled dataset because the notebook was crashing when we tried to use the entire Data set. Using the horizontal dashed red line to cut at a Euclidean distance of approximately 130 effectively partitioned the data into the five (5) distinct clusters (k=5).
- From the summary of the three (3) methods, the optimal number of Clusters chosen is k=5 for the k-means.
