## b. Tokenisation

```python
text = """
Our primary mission is to uphold the institutional commitment
to lifelong learning by providing the broadest possible access
to elite academic resources. We aim to dismantle traditional barriers
to entry, ensuring that a diverse global community can engage
with high-level scholarship regardless of their background
or previous educational experience.

We design, deliver, and oversee a comprehensive suite of research-led
academic offerings across a vast range of disciplines, including: World-class
short courses for rapid skill acquisition, Accredited Programs,By leveraging an
extensive network of internal academic expertise and wider professional affiliations,
our division serves as a primary point of entry for diverse learner pathways; Supporting
those entering higher education for the first time or re-engaging after an extended hiatus.
Delivering cutting-edge, industry-relevant curriculum for early- to mid-career professionals
and those navigating career transitions, Facilitating structured opportunities for continued
intellectual growth throughout the aging process.

We achieve our goals through a firm commitment to several operational pillars; Prioritizing affordable
tuition structure, Offering flexible modes of course delivery to suit varying lifestyles and Fostering
an inclusive atmosphere that emphasizes peer-to-peer learning and collaborative growth.
"""
```

```python
from nltk.tokenize import word_tokenize
```

```python
# Tokenisation
tokens = word_tokenize(text)
print("Tokens:")
print(tokens)
```

## c. Stopwords

```python
from nltk.corpus import stopwords
# Stopword removal
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
# Removing the stopwords
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print("Filtered Tokens:")
print(filtered_tokens)
```

## d. Removing HTML tags and punctuation

```python
import re
import string

def clean_text(text):
    # Remove HTML tags.
    text = re.sub(r'<.*?>', '', text)

    # Remove punctuation.
    text = text.translate(str.maketrans('', '', string.punctuation))

    return text
```

```python
# Clean the text
cleaned_main_text = clean_text(text)
print("Cleaned Text (main):")
print(cleaned_main_text)
```

## e. Lemmatisation

```python
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Initialize WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

# Tokenize the cleaned text
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in word_tokenize(cleaned_main_text)]

print("Lemmatized Tokens:")
print(lemmatized_tokens)
```

```python
from nltk.stem import WordNetLemmatizer
# Initialise wordnet lemmatiser.
wnl = WordNetLemmatizer()

# Perform lemmatisation.
print("{0:20}{1:20}".format("--Word--","--Lemma--"))
for word in text:
   print ("{0:20}{1:20}".format(word, wnl.lemmatize(word, pos="v")))
```

```python
# Removing Stopword from lemmised token
lemmatized_filtered_tokens = [word for word in lemmatized_tokens if word.lower() not in stop_words]
print("Lemmatized Tokens after Stopword Removal:")
print(lemmatized_filtered_tokens)
```

## f. Stemming

```python
from nltk.stem import PorterStemmer

# Initialize Porter Stemmer
porter_stemmer = PorterStemmer()

# Perform stemming on the lemmatized and filtered tokens
stemmed_tokens = [porter_stemmer.stem(word) for word in lemmatized_filtered_tokens]

print("Stemmed Tokens:")
print(stemmed_tokens)
```

```python
# Display token with their stem
for token in filtered_tokens:
    print(token + ' --> ' + porter_stemmer.stem(token))
```

```python
# Comparing the number of lemmised and stemmed token
print(f"Number of lemmatized tokens (after stopword removal): {len(lemmatized_filtered_tokens)}")
print(f"Number of stemmed tokens: {len(stemmed_tokens)}")
```

### Comparison of Original, Lemmatized, and Stemmed Tokens:

```python
import pandas as pd
from nltk.stem import WordNetLemmatizer
from nltk.stem import PorterStemmer

# Initialize lemmatizer and stemmer (ensure they are available or re-initialize)
# These objects are already initialized in previous cells, but re-initializing makes this cell self-contained.
lemmatizer = WordNetLemmatizer()
porter_stemmer = PorterStemmer()

# Create aligned lists for comparison by processing the original 'tokens' list
# This ensures all lists have the same length for direct comparison.
lemmatized_from_original_tokens = [lemmatizer.lemmatize(word) for word in tokens]
stemmed_from_original_tokens = [porter_stemmer.stem(word) for word in tokens]

# Create a DataFrame for comparison
comparison_df = pd.DataFrame({
    'Original Token': tokens,
    'Lemmatized Token': lemmatized_from_original_tokens,
    'Stemmed Token': stemmed_from_original_tokens
})

print("Comparison of Original, Lemmatized, and Stemmed Tokens:")
display(comparison_df.head(20)) # Display the first 20 for brevity, can adjust as needed
```

## g. Part-of-speech tagging

```python
# Perform speech tagging on original text
tokens = word_tokenize(text)

tagged = nltk.pos_tag(tokens)

print(tagged)
```

```python
import nltk
# Perform Part-of-Speech tagging on filtered token
pos_tags = nltk.pos_tag(lemmatized_filtered_tokens)

print("Part-of-Speech Tags:")
print(pos_tags)
```

## h. Named entity recognition



```python
# Apply named entity recognition to the tagged words.
nltk.chunk.ne_chunk(tagged)
```

```python
# Apply named entity recognition to the tagged words.
nltk.chunk.ne_chunk(pos_tags)
```
