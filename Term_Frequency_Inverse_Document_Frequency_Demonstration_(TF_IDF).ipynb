## Term Frequency - Inverse Document Frequency Representation (TF-IDF)

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer(analyzer = "word",
                             lowercase=True,
                             max_features = 50)
```

```python
docs = [
        'This is the first project.',
        'This project is the second project., the project is nice',
        'And this is the third one.',
        'Is this the first project?', ]
```

```python
# Convert the documents into a document-term matrix.
wm = vectorizer.fit_transform(docs)
print(wm.todense())
```

```python
vocabulary = vectorizer.vocabulary_
print(vocabulary)
```

```python
tokens = vectorizer.get_feature_names_out()
print(tokens)
```

```python
# Create an index for each row.
doc_names = ['Doc{:d}'.format(idx) for idx, _ in enumerate(wm)]
df_tfidf = pd.DataFrame(data=wm.toarray(), index=doc_names,
                  columns=tokens)
```
