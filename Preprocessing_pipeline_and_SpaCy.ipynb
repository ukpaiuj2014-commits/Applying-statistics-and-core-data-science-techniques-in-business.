## c. Natural language processing pipeline


The first step for a text string, when working with SpaCy, is to pass it to an NLP object. The object is a pipeline of several text pre-processing operations through which the input text string has to go through.

When invoking NLP functionalities on a text using SpaCy, the framework initially performs tokenisation to generate a `Doc` object. This object undergoes processing through multiple stages, commonly referred to as the processing pipeline. Standard pipelines in trained models usually comprise components such as a tagger, a lemmatiser, a parser, and an entity recogniser. Each component processes the `Doc` and forwards it to the subsequent component in the pipeline.

```python
text = """
Our primary mission is to uphold the institutional commitment
to lifelong learning by providing the broadest possible access
to elite academic resources. We aim to dismantle traditional barriers
to entry, ensuring that a diverse global community can engage
with high-level scholarship regardless of their background
or previous educational experience.

We design, deliver, and oversee a comprehensive suite of research-led
academic offerings across a vast range of disciplines, including: World-class
short courses for rapid skill acquisition, Accredited Programs,By leveraging an
extensive network of internal academic expertise and wider professional affiliations,
our division serves as a primary point of entry for diverse learner pathways; Supporting
those entering higher education for the first time or re-engaging after an extended hiatus.
Delivering cutting-edge, industry-relevant curriculum for early- to mid-career professionals
and those navigating career transitions, Facilitating structured opportunities for continued
intellectual growth throughout the aging process.

We achieve our goals through a firm commitment to several operational pillars; Prioritizing affordable
tuition structure, Offering flexible modes of course delivery to suit varying lifestyles and Fostering
an inclusive atmosphere that emphasizes peer-to-peer learning and collaborative growth.
"""
```

```python
doc_text = nlp(text)
```

```python
for word in doc_text:
    print(word)
```

## e. Convert the tokens into a list

```python
doc_text
```

```python
token_list = [token for token in doc_text]
```

```python
token_list
```

## f. Stopwords

Stopwords are commonly used words in a language, typically filtered out before processing text because they carry minimal unique information. Examples include words like 'the', is', 'at', 'which', and 'on'.

```python
# Remove the stopwords.
filtered_tokens = [token for token in token_list if not token.is_stop]
```

```python
filtered_tokens
```

### Note:
Although tokenisation is a kind of text normalisation, the more complex process of word normalisation requires information about the meaning which involves reducing various forms of a word to a single, standard representation. The two primary methods of word normalisation are stemming and lemmatisation.
## Lemmatisation
Lemmatisation uses a data structure that maps different forms of a word back to its simplest form, known as the lemma. Unlike stemming, lemmatisation 'understands' the context and morphological analysis of words, and accurately converts words to their base or dictionary form.

```python
for word in filtered_tokens:
  print(word.text + '  ===>', word.lemma_)
```

## h. Part-of-speech tagging
POS tag are assigned to each token depending on its usage in a sentence.

POS tags are useful for assigning a syntactic category like 'noun' or 'verb' to each word.

```python
doc
```

```python
for token in doc:
    print(f"{token.text:<10} | {token.pos_:<10} | {token.dep_:<10} | {token.head.text:<10} | {spacy.explain(token.pos_):<10}")
```

## i. Dependency parsing
Perform a critical NLP task that involves analysing the grammatical structure of a sentence to establish the relationships between 'head' words and the words that modify or are related to these heads.

SpaCy provides a highly efficient and accurate dependency parser that can be used to extract this syntactic information from text, making it easier to understand and analyse the sentence structure.

```python
# Process a sentence
doc2 = nlp(text)

# Iterate over tokens
for token in doc2:
    print(f"{token.text:<12}{token.dep_:<10}{token.head.text:<12}{token.head.pos_:<10}{[child for child in token.children]}")
```

```python
text_spans = list(doc2.sents)
displacy.render(text_spans, style="dep",jupyter=True)
```

### **Named Entity Recognition (NER)**
Named Entity Recognition (NER) is the process of locating and classifying key information—known as "entities"—within raw text. These entities are sorted into specific groups, such as individuals, companies, geographic locations, financial figures, and dates. By extracting these structured details, NER makes it easier to analyze large volumes of data, whether you're improving search engine accuracy through document tagging or automatically routing customer service requests to the right department.

### **Implementation with SpaCy**
In the SpaCy library, performing NER is straightforward:
* **The `.ents` Property:** Once a text is processed into a `Doc` object, all identified entities are stored in the `.ents` attribute.
* **Efficiency:** This property provides a quick and automated way to pull structured data out of a string of text without manual searching.

> **Key takeaway:** NER transforms messy, unstructured sentences into organized data points that software can easily interpret and act upon.

```python
# Find named entities, phrases, and concepts.
for entity in doc.ents:
    print(entity.text, entity.label_)
```

```python
displacy.render(doc, style="ent",jupyter=True)
```

#Similarity evaluation

```python
#Similarilty evaluation
doc1 = nlp("I like fast food")
doc2 = nlp("I like pizza")
doc3 = nlp("I like burger")
doc4 = nlp("I like pasta")

print(doc1.similarity(doc2))
print(doc1.similarity(doc3))
print(doc1.similarity(doc4))
```

```python
import spacy

nlp = spacy.load("en_core_web_lg")
doc1 = nlp("I missed lunch")
doc2 = nlp("I had no lunch")

doc1.similarity(doc2)
```

```python
import spacy

nlp = spacy.load("en_core_web_lg")

doc = nlp("He went to play musical instument")
doc_new = nlp("He is playing a musical device")

similarity_score = doc.similarity(doc_new)
print(f"Similarity between 'He went to play musical instument' and 'He is playing a musical device': {similarity_score}")
```
